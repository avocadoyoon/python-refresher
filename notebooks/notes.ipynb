WEEK 8: DIAGNOSTIC THINKING

1. Shapiro-Wilk
   Tests whether there is evidence against normality.

   p < .05 → evidence against normality
   p ≥ .05 → no significant evidence against normality

   IMPORTANT:
   p ≥ .05 does NOT prove that data are normally distributed.

2. Levene's test
   Tests whether group variances differ.

   p < .05 → evidence of unequal variances
   p ≥ .05 → no significant evidence of unequal variances

3. t-test
   Compares means between two independent groups.

4. Mann-Whitney U
   Non-parametric alternative for two independent groups.
   Works with ranks rather than directly comparing means.

5. Paired t-test
   Compares two measurements from the same participants.

6. Wilcoxon signed-rank
   Non-parametric alternative to the paired t-test.

7. ANOVA
   Compares means across 3+ independent groups.

8. Kruskal-Wallis
   Non-parametric alternative to one-way ANOVA.

9. Statistical significance ≠ causation.
   A significant group difference does not establish that
   group membership caused the difference.
